```python
import pandas as pd
```

```python
data = pd.read_csv('Absenteeism_data.csv')
```

```python
data.describe(include = 'all')
```

```python
raw_data = data.drop(['ID'], axis = 1)
```

```python
raw_data
```

```python
raw_data.describe()
```

```python
dummies = pd.get_dummies(raw_data['Reason for Absence'],drop_first = True)
```

```python
dummies
```

```python
reason_type_1 = dummies.loc[:, '1':'14'].max(axis=1)
reason_type_2 = dummies.loc[:, '15':'17'].max(axis=1)
reason_type_3 = dummies.loc[:, '18':'21'].max(axis=1)
reason_type_4 = dummies.loc[:, '22':].max(axis=1)
```

```python
reason_type_2
```

```python
df = pd.concat([raw_data, reason_type_1, reason_type_2, reason_type_3, reason_type_4], axis=1)
```

```python
df.columns.values
```

```python
columns_names = ['Reason for Absence', 'Date', 'Transportation Expense',
       'Distance to Work', 'Age', 'Daily Work Load Average',
       'Body Mass Index', 'Education', 'Children', 'Pets',
       'Absenteeism Time in Hours','reason_type_1','reason_type_2','reason_type_3','reason_type_4']
```

```python
df.columns = columns_names
df
```

```python
columns_names_reordered = ['reason_type_1','reason_type_2','reason_type_3','reason_type_4','Reason for Absence', 'Date', 'Transportation Expense',
       'Distance to Work', 'Age', 'Daily Work Load Average',
       'Body Mass Index', 'Education', 'Children', 'Pets',
       'Absenteeism Time in Hours']
df = df[columns_names_reordered]
df.head()
```

```python
data_with_dummies = df.drop(['Reason for Absence'], axis = 1)
```

```python
data_with_dummies
```

```python
data_with_dummies['Date'] = pd.to_datetime(data_with_dummies['Date'], dayfirst=True)
data_with_dummies['Month'] = data_with_dummies['Date'].dt.month       
data_with_dummies['Day_of_Week'] = data_with_dummies['Date'].dt.dayofweek   
data_with_dummies = data_with_dummies.drop(columns=['Date'])
```

```python
data_with_dummies
```

```python
data_with_dummies['Education'] = data_with_dummies['Education'].map(lambda x: 1 if x == 0 else 0)
```

```python
data_with_dummies
```

```python
data_with_dummies.columns.values
```

```python
columns_names_updated = ['reason_type_1', 'reason_type_2', 'reason_type_3', 'reason_type_4', 'Month','Day_of_Week',
       'Transportation Expense', 'Distance to Work', 'Age',
       'Daily Work Load Average', 'Body Mass Index', 'Education',
       'Children', 'Pets', 'Absenteeism Time in Hours']
data_with_dummies = data_with_dummies[columns_names_updated]
data_with_dummies.head()
```

```python
data_with_dummies.copy()
```

```python
data_with_dummies.to_csv('Absenteeism_preprocessed.csv', index=False)
```
