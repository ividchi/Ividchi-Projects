```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("properties.csv")

# ---------------------------
# 1. Basic Inspection
# ---------------------------
print("First 5 rows:")
print(df.head(), "\n")

print("Dataset info:")
print(df.info(), "\n")

print("Summary statistics:")
print(df.describe(include='all'), "\n")

print("Missing values:")
print(df.isnull().sum(), "\n")

print("Duplicates:")
print(df.duplicated().sum(), "\n")

# ---------------------------
# 2. Univariate Analysis
# ---------------------------
df['Size_sqft'].plot(kind='hist', bins=20, title='Distribution of Property Sizes')
plt.xlabel("Square Feet")
plt.show()

df['SalePrice'].plot(kind='hist', bins=20, title='Distribution of Sale Prices')
plt.xlabel("Sale Price")
plt.show()

df['Location'].value_counts().plot(kind='bar', title='Count of Properties by Location')
plt.ylabel("Count")
plt.show()

# ---------------------------
# 3. Bivariate Analysis
# ---------------------------
df.plot.scatter(x='Size_sqft', y='SalePrice', title='Sale Price vs Size')
plt.show()

df.boxplot(column='SalePrice', by='Location')
plt.title("Sale Price by Location")
plt.suptitle("")
plt.show()

# ---------------------------
# 4. Correlation Analysis
# ---------------------------
print("Correlation matrix:")
print(df.corr(numeric_only=True), "\n")

df.corr(numeric_only=True).plot(kind='Correlation Heatmap')
plt.show()

# ---------------------------
# 5. Group Analysis
# ---------------------------
print("Average Sale Price by Location:")
print(df.groupby('Location')['SalePrice'].mean(), "\n")

print("Average Size by Location:")
print(df.groupby('Location')['Size_sqft'].mean(), "\n")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: style (matplotlib allows color styling)
plt.style.use("seaborn-v0_8-darkgrid")

# ---------------------------------------
# 1. LOAD DATA
# ---------------------------------------
df = pd.read_csv("properties.csv")

# ---------------------------------------
# 2. BASIC INFO ABOUT THE DATA
# ---------------------------------------
print("🔹 Columns:", df.columns.tolist())
print("\n🔹 Data Types:\n", df.dtypes)
print("\n🔹 Missing Values:\n", df.isna().sum())
print("\n🔹 Duplicates:", df.duplicated().sum())
print("\n🔹 Summary Statistics:\n")
print(df.describe(include="all"))

# ---------------------------------------
# 3. DISTRIBUTION OF PROPERTY SIZE
# ---------------------------------------
plt.figure(figsize=(10,5))
plt.hist(df["Size_sqft"], bins=20, color="#4A90E2", edgecolor="black")
plt.title("Distribution of Property Size (sqft)", fontsize=14)
plt.xlabel("Size (sqft)")
plt.ylabel("Count")
plt.show()

# ---------------------------------------
# 4. DISTRIBUTION OF SALE PRICE
# ---------------------------------------
plt.figure(figsize=(10,5))
plt.hist(df["SalePrice"], bins=20, color="#50E3C2", edgecolor="black")
plt.title("Distribution of Sale Prices", fontsize=14)
plt.xlabel("Sale Price")
plt.ylabel("Count")
plt.show()

# ---------------------------------------
# 5. BOX PLOT OF SALE PRICE BY LOCATION
# ---------------------------------------
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x="Location", y="SalePrice", palette="Set2")
plt.title("Sale Price Distribution by Location", fontsize=15)
plt.xlabel("Location")
plt.ylabel("Sale Price")
plt.show()

# ---------------------------------------
# 6. RELATIONSHIP BETWEEN SIZE & SALE PRICE
# ---------------------------------------
plt.figure(figsize=(10,6))
plt.scatter(df["Size_sqft"], df["SalePrice"], 
            color="#E35050", alpha=0.7, edgecolor="black")
plt.title("Relationship Between Size (sqft) and Sale Price", fontsize=14)
plt.xlabel("Size (sqft)")
plt.ylabel("Sale Price")
plt.show()

# ---------------------------------------
# 7. CORRELATION HEATMAP
# ---------------------------------------
plt.figure(figsize=(8,5))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap", fontsize=14)
plt.show()

# ---------------------------------------
# 8. COUNT OF PROPERTIES BY LOCATION
# ---------------------------------------
plt.figure(figsize=(8,5))
df["Location"].value_counts().plot(kind="bar", color=["#4A90E2", "#50E3C2", "#E35050"])
plt.title("Count of Properties by Location", fontsize=14)
plt.xlabel("Location")
plt.ylabel("Count")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('properties.csv')

# --- Scatter Plot: Size vs Sale Price, colored by Location ---
plt.figure(figsize=(8,6))
locations = df['Location'].unique()
colors = ['red', 'blue', 'green', 'purple', 'orange', 'brown']  # distinct colors

for loc, col in zip(locations, colors):
    subset = df[df['Location'] == loc]
    plt.scatter(subset['Size_sqft'], subset['SalePrice'], label=loc, color=col)

plt.title("Property Size vs Sale Price (Colored by Location)")
plt.xlabel("Size (sqft)")
plt.ylabel("Sale Price")
plt.legend()
plt.show()

# --- Histogram 1: Sale Price distribution (10 bins) ---
plt.figure(figsize=(8,6))
plt.hist(df['SalePrice'], bins=10, color='skyblue', edgecolor='black')
plt.title("Sale Price Distribution (10 bins)")
plt.xlabel("Sale Price")
plt.ylabel("Frequency")
plt.show()

# --- Histogram 2: Size distribution (10 bins) ---
plt.figure(figsize=(8,6))
plt.hist(df['Size_sqft'], bins=10, color='salmon', edgecolor='black')
plt.title("Property Size Distribution (10 bins)")
plt.xlabel("Size (sqft)")
plt.ylabel("Frequency")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load data
df = pd.read_csv('properties.csv')

# Select meaningful numeric columns (exclude PropertyID)
numeric_df = df.select_dtypes(include='number').drop(columns=['PropertyID'], errors='ignore')

# Compute correlation matrix
corr = numeric_df.corr().round(3)

# Plot heatmap with coolwarm colormap
plt.figure(figsize=(6,5))
plt.imshow(corr, interpolation='nearest', cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.xticks(ticks=np.arange(len(corr.columns)), labels=corr.columns, rotation=45)
plt.yticks(ticks=np.arange(len(corr.columns)), labels=corr.columns)

# Annotate values
for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        plt.text(j, i, corr.iloc[i, j], ha='center', va='center')

plt.colorbar()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load data
df = pd.read_csv('properties.csv')

# Numeric columns excluding PropertyID
numeric_df = df.select_dtypes(include='number').drop(columns=['PropertyID'], errors='ignore')

# Create single-page dashboard
fig, axes = plt.subplots(2, 3, figsize=(18, 12))

# 1. Histogram of SalePrice (10 bins)
axes[0, 0].hist(numeric_df['SalePrice'], bins=10)
axes[0, 0].set_title("Sale Price Distribution (10 bins)")
axes[0, 0].set_xlabel("Sale Price")
axes[0, 0].set_ylabel("Frequency")

# 2. Histogram of Size (10 bins)
axes[0, 1].hist(numeric_df['Size_sqft'], bins=10)
axes[0, 1].set_title("Size Distribution (10 bins)")
axes[0, 1].set_xlabel("Size (sqft)")
axes[0, 1].set_ylabel("Frequency")

# 3. Scatter: Size vs price
axes[0, 2].scatter(numeric_df['Size_sqft'], numeric_df['SalePrice'])
axes[0, 2].set_title("Size vs Sale Price")
axes[0, 2].set_xlabel("Size (sqft)")
axes[0, 2].set_ylabel("Sale Price")

# 4. Boxplot: Sale price by location
df.boxplot(column='SalePrice', by='Location', ax=axes[1, 0])
axes[1, 0].set_title("SalePrice by Location")
axes[1, 0].set_xlabel("Location")
axes[1, 0].set_ylabel("Sale Price")

# 5. Correlation heatmap
corr = numeric_df.corr()
im = axes[1, 1].imshow(corr, interpolation='nearest')
axes[1, 1].set_title("Correlation Heatmap")
axes[1, 1].set_xticks(np.arange(len(corr.columns)))
axes[1, 1].set_xticklabels(corr.columns, rotation=45)
axes[1, 1].set_yticks(np.arange(len(corr.columns)))
axes[1, 1].set_yticklabels(corr.columns)
plt.colorbar(im, ax=axes[1, 1])

# 6. Line plot: Price trend over sorted size
sorted_df = df.sort_values('Size_sqft')
axes[1, 2].plot(sorted_df['Size_sqft'], sorted_df['SalePrice'])
axes[1, 2].set_title("Price Trend by Size")
axes[1, 2].set_xlabel("Size (sqft)")
axes[1, 2].set_ylabel("Sale Price")

plt.tight_layout()
plt.show()
```
